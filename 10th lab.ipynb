```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns 
import sklearn
from sklearn.cluster import KMeans
```

```python
sns.set()
```

```python
#load the data
data=pd.read_csv("C:\\Users\\paswa\\Downloads\\archive\\3.01.Countryclusters.csv")
data
```

```python
#plotting the dataset
plt.scatter(data['Latitude'],data['Longitude'])
plt.xlim(-180,180)
plt.ylim(-90,90)
plt.show()
```

```python
#SELECTING THE FEATURE
x=data.iloc[:,1:3]
x
```

```python
#CLUSTERING 
Kmeans=KMeans(5)
Kmeans.fit(x)
#CLUSTERING RESULTS 
identified_clusters=Kmeans.fit_predict(x)
#the identified clusters where our data can go
identified_clusters
#creating a copy of our data 
data_with_clusters=data.copy()
#new clustering -----> data
data_with_clusters['Clusters']=identified_clusters
#plot the data
plt.scatter(data_with_clusters['Latitude'],data_with_clusters['Longitude'],c=data_with_clusters['Clusters'],cmap='rainbow')
plt.show()
```

```python
#write a pyt program to demonstrate K-means custering data set: (10,2), 12,4:: 16,6::20,8::24,10::26,12::30,14::34,16
import math

# Dataset
data = [
    (10, 2),
    (12, 4),
    (16, 6),
    (20, 8),
    (24, 10),
    (26, 12),
    (30, 14),
    (34, 16)
]

# Initial centroids
c1 = (10, 2)
c2 = (34, 16)

for iteration in range(5):

    cluster1 = []
    cluster2 = []

   
    for point in data:
        d1 = math.sqrt((point[0]-c1[0])**2 +
                       (point[1]-c1[1])**2)

        d2 = math.sqrt((point[0]-c2[0])**2 +
                       (point[1]-c2[1])**2)

        if d1 < d2:
            cluster1.append(point)
        else:
            cluster2.append(point)

    
    c1 = (
        sum(p[0] for p in cluster1) / len(cluster1),
        sum(p[1] for p in cluster1) / len(cluster1)
    )

    c2 = (
        sum(p[0] for p in cluster2) / len(cluster2),
        sum(p[1] for p in cluster2) / len(cluster2)
    )

    print("Iteration:", iteration + 1)
    print("Cluster 1:", cluster1)
    print("Cluster 2:", cluster2)
    print("Centroid 1:", c1)
    print("Centroid 2:", c2)
    print()
```
